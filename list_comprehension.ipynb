```python
prices = [250, 899, 45, 1500, 620, 30, 999]

tax_added = list(map(lambda x:x*0.10 ,prices))

new_lst = list(filter(lambda x : x>500 , prices))

print(tax_added)
print(new_lst)
```

```python
from functools import reduce

sum = reduce(lambda x,y : x+y , prices)

print(sum)
```

```python
avg = (reduce(lambda x,y: x+y ,new_lst))/len(new_lst)
avg
```

```python
nums = [4, 9, 15, 22, 31, 40, 55]

div_5 = [x for x in nums if x%5 == 0 ]
div_5
```
