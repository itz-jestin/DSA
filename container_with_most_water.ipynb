```python
def most_water(cont_list):
  max_area = 0
  area = 0
  for i in range(len(cont_list)):
    for j in range(i+1,len(cont_list)):
      width= j - i
      area = min(cont_list[i],cont_list[j]) * width
      if area>max_area:
        max_area = area
  return (max_area)

most_water([1,8,6,2,5,4,8,3,7])
```
