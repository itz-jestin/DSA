```python
def most_water(cont_list):
  max_area = 0
  area = 0
  for i in range(len(cont_list)):
    for j in range(i+1,len(cont_list)):
      width= j - i
      area = min(cont_list[i],cont_list[j]) * width
      if area>max_area:
        max_area = area
  return (max_area)

most_water([1,8,6,2,5,4,8,3,7])
```

```python
def most_water(cont_list):
  max_area = 0
  area = 0
  left = 0
  right = len(cont_list)-1
  while left<right:
    width = right - left
    area = min(cont_list[left],cont_list[right])*width
    if cont_list[left]<cont_list[right]:
      left+=1
    else:
      right-=1
    if area>max_area:
      max_area = area
  return max_area


most_water([1,8,6,2,5,4,8,3,7])
```
