Write a menu-driven program that chooses a function dynamically from a dictionary based on user input.

```python
def add(a,b):
  return a+b
def sub(a,b):
  return a-b
menu={
    "+":add,
    "-":sub,
}
a=int(input("enter a digit "))
b=int(input("enter another digit "))
user_in=input("Enter the operation to do ")

print(f"{menu[user_in]} of {a} and {b} is :{menu[user_in](a,b)}")
```
