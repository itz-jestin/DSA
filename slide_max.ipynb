```python
from collections import deque

class MovingAverage:
  def __init__(self,k):
    self.k = k
    self.window = deque()

  def next(self,val):
    self.window.append(val)
    if len(self.window)>self.k:
      self.window.popleft()
    avg=sum(self.window)/len(self.window)
    return avg
```

```python
ma = MovingAverage(3)
print(ma.next(1))    # 1.0
print(ma.next(10))   # 5.5
print(ma.next(3))    # 4.666...
print(ma.next(5))    # 6.0
```

```python
from collections import deque

class SlideMax:
  def __init__(self,k):
    self.k = k
    self.window = deque()

  def next(self,val):
    self.window.append(val)
    if len(self.window) > self.k:
      self.window.popleft()
    if len(self.window) == self.k:
      return max(self.window)
    else:
      return None

def SlideMaxRun(nums,k):
  current_num=SlideMax(k)
  arr=[]
  for elem in nums:
    res = current_num.next(elem)
    if res is not None:
      arr.append(res)
  return arr
print(SlideMaxRun([1,3,-1,-3,5,3,6,7],3))
```
