```python
def replace(lst):
  i = 0
  j=i+1
  while j<len(lst):
    if lst[i]==0 and lst[j]!=0:
      lst[i]=lst[j]
      lst[j]=0
      i+=1
      j+=1
    elif lst[i]==0 and lst[j]==0:
      lst[i]=lst[j]
      lst[j]=0
      j+=1
    else:
      i+=1
      j+=1
  return lst
print(replace([0, 0, 0, 0, 12]))

```
