```python
def min_win_sub(s,t):
  need={}
  for c in t:
    need[c]=need.get(c,0)+1
  print("need",need)

  window_count={}
  left=0
  formed=0
  min_wind=float('inf')

  for right in range(len(s)):
    char=s[right]
    if char in need:
      window_count[char]=window_count.get(char,0)+1

      if window_count[char]==need[char]:  #it checks whether need is fullfilled or not
        formed+=1
    else:
      continue
    while formed == 3:
      min_wind = min(min_wind, right-left+1)

      left_char = s[left]
      if left_char in need:
        window_count[left_char]-=1
        if window_count[left_char] < need[left_char]:
          formed -=1

      left +=1
    print("left",left,"right",right,"formed:",formed,":",s[left:right+1],":","current char-",s[right],window_count,min_wind)
  return min_wind
s = "ADOBECODEBANC"
t = "ABC"
min_win_sub(s,t)
```
