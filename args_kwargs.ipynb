```python
def build_profile(name,*hobbies,**user_info):
  return {
      "name":name,
      "hobbies": list(hobbies),
      "details": user_info
  }
print(build_profile("Jestin","chess","reading",age=30,city="Kochi"))
```

```python
def order_pizza(size, *toppings, **extras):
    print(f"Size: {size}")
    print(f"Toppings: {toppings}")
    print(f"Extras: {extras}")

size = "Large"
toppings_list = ["mushroom", "olives", "onion"]
extra_info = {"crust": "thin", "delivery": True}

order_pizza(size,*toppings_list,**extra_info)
```

```python
def make_sandwich(bread, *fillings, toasted, **extras):
    print(f"Bread: {bread}")
    print(f"Fillings: {fillings}")
    print(f"Toasted: {toasted}")
    print(f"Extras: {extras}")
bread = "Sourdough"
fillings_list = ["turkey", "lettuce", "tomato"]
other_stuff = {"sauce": "mustard", "cheese": "cheddar"}

make_sandwich(bread,*fillings_list,toasted=True,**other_stuff)
```
