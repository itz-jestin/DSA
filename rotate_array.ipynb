```python
def rotate_array(nums,k=1):
  if k<len(nums):
    dup1=nums[k+1:len(nums)]
    dup2=nums[0:k+1]
    print(dup1,dup2)
    nums[0:k+1]=dup1
    nums[k:len(nums)] = dup2
  else:
    k=k % len(nums)
    dup1=nums[k+1:len(nums)]
    dup2=nums[0:k+1]
    print(dup1,dup2)
    nums[0:k+1]=dup1
    nums[k:len(nums)] = dup2

  return nums
print(rotate_array([1,2],3))
```
