longest substring without repeating characters

```python
def long_sub(string):
  count={}
  str_new = ""
  i=0
  n=len(string)
  while i<n:
    if string[i] not in str_new:
      str_new+=string[i]
      i+=1
    i+=1
  return str_new
print(long_sub("aabbo"))
```

```python
def long_sub1(string):
  curr_word=0
  max_count=1
  str_new=""
  count=0
  n=len(string)
  while curr_word<n:
    if string[curr_word] not in str_new:
      str_new+=string[curr_word]
      curr_word+=1
      count+=1
    else:
      str_new=""
      if count>max_count:
        max_count=count
        count=0
      else:
        count=0
  return max_count
print("Longest substring without repeats",long_sub1("abcabcbb"))
```

```python
def long_sub2(string):
  left=0
  right=0
  max_length=0
  seen=set()
  while right<len(string):
    while string[right] in seen:
      seen.remove(string[left])
      left+=1
    seen.add(string[right])
    curr_window_size=right-left+1
    if curr_window_size>max_length:
      max_length=curr_window_size
    right+=1
  return seen ,max_length
print(long_sub2("abbc"))
```
