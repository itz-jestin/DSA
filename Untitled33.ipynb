longest substring without repeating characters

```python
def long_sub(string):
  count={}
  str_new = ""
  i=0
  n=len(string)
  while i<n:
    if string[i] not in str_new:
      str_new+=string[i]
      i+=1
    i+=1
  return str_new
print(long_sub("aabbo"))
```
