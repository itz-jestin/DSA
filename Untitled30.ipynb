```python
def dutch_flag_prob(lst):
  low = 0
  mid = 0
  high=len(lst)-1
  a = 0
  while mid<=high:
    if lst[mid]==0:
      a=lst[mid]
      lst[mid]=lst[low]
      lst[low]=a
      a=0
      low+=1
      mid+=1
    elif lst[mid]==1:
      mid+=1
    elif lst[mid]==2:
      a=lst[high]
      lst[high]=lst[mid]
      lst[mid]=a
      a=0
      high-=1
    return lst
print(dutch_flag_prob([2, 0, 2, 1, 1, 0]))
```
