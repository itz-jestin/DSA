```python
def rem_dup(lst):
  seen = set()
  dup = set()
  for i in range(len(lst)):
    if lst[i] in seen:
      dup.add(lst[i])
    else:
      seen.add(lst[i])
  return list(seen)
num_list = [0,0,1,1,1,2,2,3,3,4]
print(rem_dup(num_list))
```

```python
def rem_dup2(lst):
  insert_pos = 1
  i = 1
  while i<len(lst):
    if lst[i]!=lst[insert_pos-1]:
      lst[insert_pos]=lst[i]
      insert_pos+=1
      i+=1
    else:
      i+=1
  return insert_pos
print(rem_dup2(num_list))
print(num_list)
```
