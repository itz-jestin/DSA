```python
def trap_water(water_list):
  n=len(water_list)
  left = 0
  right = n-1
  i = 1
  total_stored = 0
  stored=0
  while i<=n-2:
    left_sort = water_list[left:i]
    left_max = max(left_sort)
    right_sort = water_list[i+1:right+1]
    right_max = max(right_sort)
    stored = min(left_max,right_max) - water_list[i]
    if stored<0:
      stored=0
    total_stored += stored
    i+=1
  return total_stored

trap_water([0,1,0,2,1,0,1,3,2,1,2,1])
```
