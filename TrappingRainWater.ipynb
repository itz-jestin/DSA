```python
def trap_water(water_list):
  n=len(water_list)
  left = 0
  right = 2
  i = 1
  max_stored = 0
  while right<=n-1:
    water_stored = min(water_list[left],water_list[right]) - water_list[i]
    left +=1
    right+=1
    i+=1
    if water_stored > max_stored:
      max_stored = water_stored
    return(water_stored)
trap_water([3, 0, 2])
```
