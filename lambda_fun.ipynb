```python
lst = [1,2,3,4,5,6]

square=list(map(lambda x : x*x ,lst))
print(square)
```

```python
names = ["john","alice","bob"]
last_n = ["ahe","die","wohg"]
cap = list(map(lambda x,y : f"{x} {y}",names,last_n))
print(cap)
```
