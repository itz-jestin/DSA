```python
def three_sum(nums):
  nums.sort()
  result=[]
  dup=[]
  for i in range(len(nums)):
    for j in range(len(nums)):
      for k in range(len(nums)):
        if i!=j and i!=k and j!=k:
          if nums[i]+nums[j]+nums[k]==0:
            sorted_array=[nums[i],nums[j],nums[k]]
            sorted_array.sort()
            result.append(sorted_array) if (sorted_array) not in result else dup.append(sorted_array)
  return result
nums=[-1, 0, 1, 2, -1, -4]
print(three_sum(nums))
```

```python
def three_sum(nums):
  nums.sort()
  n = len(nums)
  result = []
  dup=[]
  for i in range(n):
    left = i+1
    right = n-1
    while left < right:
        if nums[i]+nums[left]+nums[right]==0:
          sorted_array=[nums[i],nums[left],nums[right]]
          sorted_array.sort()
          result.append(sorted_array) if (sorted_array) not in result else dup.append(sorted_array)
    return result
nums=[-1, 0, 1, 2, -1, -4]
print(three_sum(nums))

```
