```python
def prod(nums):
  left_prod = []
  j=-1
  right_prod = []
  for i in range(0,len(nums)):
    if left_prod or right_prod:
      left_prod[i]=(left_prod[-1]*nums[i])
      right_prod.append(right_prod[j])
      j-=1
    else:
      left_prod.append(1*nums[i])
      right_prod.append(1*nums[-1])
      j-=1

  return left_prod
print(prod([2,2,3,4]))
```

```python
def prod(nums):
  n=len(nums)
  total=[1]*n
  left_prod = [1]*n
  right_prod = [1]*n
  for i in range(1,n):
    left_prod[i]=left_prod[i-1]*nums[i-1]
    right_prod[-1-i]=right_prod[-i]*nums[-i]

  for j in range(n):
    total[j]=left_prod[j]*right_prod[j]
  return total
print(prod([2,3,4,5]))
```
