```python
def prod(nums):
  left_prod = []
  j=-1
  right_prod = []
  for i in range(0,len(nums)):
    if left_prod or right_prod:
      left_prod[i]=(left_prod[-1]*nums[i])
      right_prod.append(right_prod[])
    else:
      left_prod.append(1*nums[i])
      right_prod.append(1*nums[-1])
      j-=1

  return left_prod
print(prod([2,2,3,4]))
```
