```python
def merge_arr(nums1,nums2):
  left = 0
  nums2_pointer = 0
  while left<len(nums1):
    if nums1[left] == 0:
      nums1[left]=nums2[nums2_pointer]
      left+=1
      nums2_pointer +=1
    else:
      left+=1
  return nums1

print(merge_arr( [1, 2, 3, 0, 0, 0],[2, 5, 6]))
```
