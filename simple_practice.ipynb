```python
def reverse(string):
  rev=""
  for i in range(len(string)-1,-1,-1):
    rev+=string[i]
  return rev
print(reverse("hello"))
```

```python
def format(input):
  rev=""
  for i in range(len(input)):
    rev+=input[i]
  return(rev)
print(format("{}hello"))
```
