```python
def reverse(string):
  rev=""
  for i in range(len(string)-1,-1,-1):
    rev+=string[i]
  return rev
print(reverse("hello"))
```

```python

def string_format(input):
  stack=[]
  popped = ""
  if len(input)==1:
    return False
  for i in range(len(input)):
    if input[i]=="(" or input[i]=="[" or input[i]=="{" :
      stack.append(input[i])
    else:
      if stack==[]:
        return False
      popped = stack.pop()
      if popped=="(" and input[i]==")":
        continue
      elif popped=="[" and input[i]=="]":
        continue
      elif popped=="{" and input[i]=="}":
        continue
      else:
        return False

  if stack==[]:
    return True
  else:
    return False

string_format("{{]}")
```
