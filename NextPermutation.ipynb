```python
def next_per1(nums):
  r_place = len(nums)-1
  side = r_place - 1
  while side!=0:
    if nums[side]<nums[r_place]:
      a = nums[side]
      nums[side] = nums[r_place]
      nums[r_place] = a
      r_place-=1
      side-=1
    else:
      r_place-=1
      side-=1
  return nums
print(next_per1([1,2,3,4]))
```
