```python
def min_wind_sub(s,t):
  left=0
  count={}
  total=0
  min_len=0
  for right in range(len(s)):
    if s[right] in t:
      count[s[right]]=count.get(s[right,0])+1
    for j in (t):
      if count[j]==1:
        total+=1
    if total==3:
      min_len=min(right-left+1,min_len)

    elif right<len(s):
      return {}

min_wind_sub("ABC","CD")

```
