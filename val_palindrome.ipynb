Valid Palindrome



```python
def val_palindrome(string):
  string=string.lower()
  # symbols = "(%<>,;:) "
  new=""
  for w in string:
    if w.isalnum():
      new+=w
  rev_sent=new[::-1]
  return "Palindrome" if new==rev_sent else "Not Palindrome"
print(val_palindrome("A man, a plan, a canal: Panama"))
```
