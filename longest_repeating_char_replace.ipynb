```python
def long_rep(string):
  right=0
  left=0
  count_dict={}
  while right<len(string):
    if string[right] in count_dict:
      count_dict[string[right]]+=1
    else:
      count_dict[string[right]]=1

    right+=1
  return count_dict
print(long_rep("ABAC"))
```

```python
def long_rep1(string,k=1):
  s=string.upper()
  left=0
  count_dic=dict()
  max_length=0
  max_freq=0
  for right in range(len(s)):
    count_dic[s[right]]=count_dic.get(s[right],0)+1
    max_freq = max(max_freq,count_dic[s[right]])

    while (right-left+1)-max_freq > k:
      count_dic[s[left]]-=1
      left+=1
    max_length=max(max_length,right-left+1)
  return max_length
long_rep1("ABAABE",1)
```
