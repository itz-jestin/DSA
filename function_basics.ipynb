```python
def larg(a,b):
  if a>b:
    print(f"{a} is greater")
  elif(a<b):
    print(f"{b} is greater")
  elif(a==b):
    print("Same no.")
  else:
    print("Not a no.")

larg(10,3)
```

```python
def calc(a,b):
  sum = a+b
  prod = a*b
  avg = (a+b)/2
  return sum,prod,avg
sum,prod,avg=calc(2,2)
print(f"sum {sum},product {prod},avg {avg}")
```

```python

def prime(n):
  arr=[]
  for i in range(1,n+1):
    print(i)
    if n%i==0 and n%1==0:
      arr.append(i)
  print(f"numbers {arr}")
prime(20)
```
